### Converting data types to their correct formats.
Numeric columns (Quantity, Price, Total) were converted to numeric types, and Transaction Date was converted to datetime.
The errors='coerce' option ensures invalid entries are turned into NaN for later handling.



```python
df['Quantity'] = pd.to_numeric(df['Quantity'],errors = 'coerce')

df['Price Per Unit'] = pd.to_numeric(df['Price Per Unit'],errors = 'coerce')

df['Total Spent'] = pd.to_numeric(df['Total Spent'],errors = 'coerce')

df['Transaction Date'] = pd.to_datetime(df['Transaction Date'], format='%Y-%m-%d', errors='coerce')
```

##### Handling Missing Values Using Logical Relationships

```python
data = pd.DataFrame(df)
```

```python
for x in data.index:
    if str(data.loc[x,'Total Spent']) == 'nan':
        data.loc[x,'Total Spent']= float(data.loc[x,'Quantity']*data.loc[x,'Price Per Unit'])
```

```python
data['Total Spent'].isna().sum()
```

```python
pd.DataFrame(data.groupby('Item')['Price Per Unit'].unique())
```

```python
item = {'Cake':3.0,'Coffee':2.0,'Cookie':1.0,'Juice':3.0,'Salad':5.0,'Sandwich':4.0,'Smoothie':4.0,'Tea':1.5}
```

```python
item.keys()
```

```python
for i in data.index:
    if str(data.loc[i,'Price Per Unit']) =='nan' and data.loc[i,'Item'] in item.keys():
        data.loc[i,'Price Per Unit']= float(item[str(data.loc[i,'Item'])])
```

```python
data['Price Per Unit'].isna().sum()
```

##### In this step, I will handle missing values by recalculating them using existing relationships between columns:

If Total Spent is missing → I will compute it as:
Total Spent = Quantity × Price Per Unit

If Price Per Unit is missing → I will compute it as:
Price Per Unit = Total Spent ÷ Quantity

If Quantity is missing → I will compute it as:
Quantity = Total Spent ÷ Price Per Unit

After applying these formulas, I will re-check for any remaining missing values.
Rows with incomplete or illogical data will be dropped to maintain dataset quality.

```python
for i in data.index:
    if str(data.loc[i,'Price Per Unit']) == 'nan':
        data.loc[i,'Price Per Unit'] = float(data.loc[i,'Total Spent'])/float(data.loc[i,'Quantity'])
```

```python
data['Price Per Unit'].isna().sum()
```

```python
for x in data.index:
    if str(data.loc[x,'Total Spent']) == 'nan':
        data.loc[x,'Total Spent']= float(data.loc[x,'Quantity']*data.loc[x,'Price Per Unit'])
```

```python
data['Total Spent'].isna().sum()
```

```python
for i in data.index: 
    if str(data.loc[i,'Quantity']) == 'nan' :
        data.loc[i,'Quantity'] = float(data.loc[i,'Total Spent']/data.loc[i,'Price Per Unit'])
```

```python
data['Quantity'].isna().sum()
```

```python
data[['Quantity','Price Per Unit','Total Spent']].isna().sum()
```

```python
data[['Quantity', 'Total Spent', 'Price Per Unit']].isna().any(axis=1).sum()
```
